```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
import time

print("Libraries imported successfully!")
```

```python
from google.colab import files

uploaded = files.upload()
IMAGE_PATH = list(uploaded.keys())[0]

img = cv2.imread(IMAGE_PATH)

if img is None:
    print("Error: Image could not be loaded.")
else:
    print("Image loaded successfully!")
    print("Image file:", IMAGE_PATH)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

print("Image shape:", gray.shape)
print("Image data type:", gray.dtype)

plt.figure(figsize=(8, 5))
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

```python
gaussian = cv2.GaussianBlur(gray, (5, 5), 0)

plt.figure(figsize=(8, 5))
plt.imshow(gaussian, cmap="gray")
plt.title("Gaussian Blur")
plt.axis("off")
plt.show()
```

```python
median = cv2.medianBlur(gray, 5)

plt.figure(figsize=(8, 5))
plt.imshow(median, cmap="gray")
plt.title("Median Filtering")
plt.axis("off")
plt.show()
```

```python
average = cv2.blur(gray, (5, 5))

plt.figure(figsize=(8, 5))
plt.imshow(average, cmap="gray")
plt.title("Average (Mean) Filtering")
plt.axis("off")
plt.show()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(18, 5))

axes[0].imshow(gray, cmap="gray")
axes[0].set_title("Original")

axes[1].imshow(gaussian, cmap="gray")
axes[1].set_title("Gaussian Blur")

axes[2].imshow(median, cmap="gray")
axes[2].set_title("Median Filter")

axes[3].imshow(average, cmap="gray")
axes[3].set_title("Average Filter")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

```python
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian_abs = cv2.convertScaleAbs(laplacian)

plt.figure(figsize=(8, 5))
plt.imshow(laplacian_abs, cmap="gray")
plt.title("Laplacian Filtering")
plt.axis("off")
plt.show()
```

```python
sobel_x = cv2.Sobel(gray, cv2.CV_64F, 1, 0, ksize=3)
sobel_x_abs = cv2.convertScaleAbs(sobel_x)

plt.figure(figsize=(8, 5))
plt.imshow(sobel_x_abs, cmap="gray")
plt.title("Sobel X - Horizontal Edges")
plt.axis("off")
plt.show()
```

```python
sobel_y = cv2.Sobel(gray, cv2.CV_64F, 0, 1, ksize=3)
sobel_y_abs = cv2.convertScaleAbs(sobel_y)

plt.figure(figsize=(8, 5))
plt.imshow(sobel_y_abs, cmap="gray")
plt.title("Sobel Y - Vertical Edges")
plt.axis("off")
plt.show()
```

```python
fig, axes = plt.subplots(1, 4, figsize=(18, 5))

axes[0].imshow(gray, cmap="gray")
axes[0].set_title("Original")

axes[1].imshow(laplacian_abs, cmap="gray")
axes[1].set_title("Laplacian")

axes[2].imshow(sobel_x_abs, cmap="gray")
axes[2].set_title("Sobel X")

axes[3].imshow(sobel_y_abs, cmap="gray")
axes[3].set_title("Sobel Y")

for ax in axes:
    ax.axis("off")

plt.tight_layout()
plt.show()
```
